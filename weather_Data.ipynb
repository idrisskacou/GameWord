```python
import psycopg2
import random
import datetime
```

```python
# PostgreSQL database connection details
conn = psycopg2.connect(
    dbname="mathphysic",
    user="postgres",
    password="1234",
    host="localhost",
    port="5432"
)
```

```python
cur = conn.cursor()
```

```python

# SQL statement to create the table
create_table_query = """
CREATE TABLE IF NOT EXISTS launch_data (
    id SERIAL PRIMARY KEY,
    launch_id UUID REFERENCES launch(id),
    timestamp TIMESTAMP NOT NULL,
    surface_temperature FLOAT,
    surface_wind_speed FLOAT,
    surface_wind_direction VARCHAR(2),
    surface_humidity INTEGER,
    precipitation_type VARCHAR(10),
    precipitation_intensity FLOAT,
    visibility FLOAT,
    upper_wind_speed_0_1km FLOAT,
    upper_wind_speed_1_5km FLOAT,
    upper_wind_speed_5_10km FLOAT,
    upper_wind_direction_0_1km VARCHAR(2),
    upper_wind_direction_1_5km VARCHAR(2),
    upper_wind_direction_5_10km VARCHAR(2),
    temperature_profile_0_1km FLOAT,
    temperature_profile_1_5km FLOAT,
    temperature_profile_5_10km FLOAT,
    humidity_profile_0_1km INTEGER,
    humidity_profile_1_5km INTEGER,
    humidity_profile_5_10km INTEGER,
    pressure_profile_0_1km INTEGER,
    pressure_profile_1_5km INTEGER,
    pressure_profile_5_10km INTEGER,
    turbulence_level_0_1km VARCHAR(10),
    turbulence_level_1_5km VARCHAR(10),
    turbulence_level_5_10km VARCHAR(10),
    solar_radiation_levels FLOAT,
    geomagnetic_activity INTEGER,
    solar_flare_activity VARCHAR(10),
    cosmic_ray_intensity FLOAT,
    lightning_frequency INTEGER,
    lightning_proximity FLOAT,
    cloud_base_altitude FLOAT,
    cloud_top_altitude FLOAT,
    cloud_thickness FLOAT,
    cloud_coverage_percentage INTEGER,
    sea_surface_temperature FLOAT,
    wave_height FLOAT,
    wave_direction VARCHAR(2),
    current_speed FLOAT,
    current_direction VARCHAR(2),
    pm2_5 FLOAT,
    pm10 FLOAT,
    co FLOAT,
    o3 FLOAT,
    no2 FLOAT,
    so2 FLOAT,
    ground_temperature FLOAT,
    air_temperature_0_1km FLOAT,
    air_temperature_1_5km FLOAT,
    air_temperature_5_10km FLOAT,
    dew_point FLOAT,
    ground_moisture_level INTEGER,
    soil_temperature FLOAT,
    fog_presence_and_density VARCHAR(20)
);
"""
# Execute creation of table the SQL statement 
cur.execute(create_table_query)
conn.commit()
```

```python

# Function to generate random data for a given field
def generate_random_data(base_date):
    days_offset = random.randint(-3, 3)  # Randomly select days between -3 and 3
    random_timestamp = base_date + datetime.timedelta(days=days_offset)
    
def generate_random_data(base_date):
    days_offset = random.randint(-3, 3)  # Randomly select days between -3 and 3
    random_timestamp = base_date + datetime.timedelta(days=days_offset)
    data = {
        "timestamp": random_timestamp,
        "weather_conditions": {
            "surface_temperature": round(random.uniform(15.0, 35.0), 1),
            "surface_wind_speed": round(random.uniform(0.0, 50.0), 1),
            "surface_wind_direction": random.choice(["N", "NE", "E", "SE", "S", "SW", "W", "NW"]),
            "surface_humidity": random.randint(20, 100),
            "precipitation_type": random.choice(["none", "rain", "snow"]),
            "precipitation_intensity": round(random.uniform(0.0, 50.0), 1),
            "visibility": round(random.uniform(1.0, 20.0), 1)
        },
        "upper_atmosphere_conditions": {
            "upper_wind_speeds": {
                "0-1km": round(random.uniform(0.0, 50.0), 1),
                "1-5km": round(random.uniform(0.0, 50.0), 1),
                "5-10km": round(random.uniform(0.0, 50.0), 1)
            },
            "upper_wind_directions": {
                "0-1km": random.choice(["N", "NE", "E", "SE", "S", "SW", "W", "NW"]),
                "1-5km": random.choice(["N", "NE", "E", "SE", "S", "SW", "W", "NW"]),
                "5-10km": random.choice(["N", "NE", "E", "SE", "S", "SW", "W", "NW"])
            },
            "temperature_profiles": {
                "0-1km": round(random.uniform(-20.0, 20.0), 1),
                "1-5km": round(random.uniform(-60.0, 0.0), 1),
                "5-10km": round(random.uniform(-70.0, -10.0), 1)
            },
            "humidity_profiles": {
                "0-1km": random.randint(10, 100),
                "1-5km": random.randint(10, 100),
                "5-10km": random.randint(10, 100)
            },
            "pressure_profiles": {
                "0-1km": random.randint(900, 1100),
                "1-5km": random.randint(400, 900),
                "5-10km": random.randint(100, 400)
            },
            "turbulence_levels": {
                "0-1km": random.choice(["low", "medium", "high"]),
                "1-5km": random.choice(["low", "medium", "high"]),
                "5-10km": random.choice(["low", "medium", "high"])
            }
        },
        "solar_activity": {
            "solar_radiation_levels": round(random.uniform(1200, 1400), 1),
            "geomagnetic_activity": random.randint(0, 9),
            "solar_flare_activity": random.choice(["none", "low", "moderate", "high"]),
            "cosmic_ray_intensity": round(random.uniform(2.0, 6.0), 1)
        },
        "lightning_activity": {
            "lightning_frequency": random.randint(0, 100),
            "lightning_proximity": round(random.uniform(0.0, 100.0), 1)
        },
        "cloud_cover": {
            "cloud_base_altitude": round(random.uniform(0.5, 10.0), 1),
            "cloud_top_altitude": round(random.uniform(1.0, 15.0), 1),
            "cloud_thickness": round(random.uniform(0.1, 15.0), 1),
            "cloud_coverage_percentage": random.randint(0, 100)
        },
        "marine_conditions": {
            "sea_surface_temperature": round(random.uniform(15.0, 30.0), 1),
            "wave_height": round(random.uniform(0.0, 10.0), 1),
            "wave_direction": random.choice(["N", "NE", "E", "SE", "S", "SW", "W", "NW"]),
            "current_speed": round(random.uniform(0.0, 3.0), 1),
            "current_direction": random.choice(["N", "NE", "E", "SE", "S", "SW", "W", "NW"])
        },
        "air_quality": {
            "pm2_5": round(random.uniform(0.0, 50.0), 1),
            "pm10": round(random.uniform(0.0, 100.0), 1),
            "co": round(random.uniform(0.0, 10.0), 1),
            "o3": round(random.uniform(0.0, 100.0), 1),
            "no2": round(random.uniform(0.0, 50.0), 1),
            "so2": round(random.uniform(0.0, 20.0), 1)
        },
        "thermal_conditions": {
            "ground_temperature": round(random.uniform(15.0, 35.0), 1),
            "air_temperature_at_different_altitudes": {
                "0-1km": round(random.uniform(15.0, 35.0), 1),
                "1-5km": round(random.uniform(-20.0, 0.0), 1),
                "5-10km": round(random.uniform(-70.0, -10.0), 1)
            },
            "dew_point": round(random.uniform(0.0, 25.0), 1)
        },
        "launch_site_specific_conditions": {
            "ground_moisture_level": random.randint(0, 100),
            "soil_temperature": round(random.uniform(15.0, 35.0), 1),
            "fog_presence_and_density": random.choice(["none", "light", "moderate", "dense"])
        }
    }
    return data
```

```python

# Function to insert generated data into the database
def insert_data(launch_id, data):
    query = """
    INSERT INTO launch_data (
        launch_id,
        timestamp,
        surface_temperature,
        surface_wind_speed,
        surface_wind_direction,
        surface_humidity,
        precipitation_type,
        precipitation_intensity,
        visibility,
        upper_wind_speed_0_1km,
        upper_wind_speed_1_5km,
        upper_wind_speed_5_10km,
        upper_wind_direction_0_1km,
        upper_wind_direction_1_5km,
        upper_wind_direction_5_10km,
        temperature_profile_0_1km,
        temperature_profile_1_5km,
        temperature_profile_5_10km,
        humidity_profile_0_1km,
        humidity_profile_1_5km,
        humidity_profile_5_10km,
        pressure_profile_0_1km,
        pressure_profile_1_5km,
        pressure_profile_5_10km,
        turbulence_level_0_1km,
        turbulence_level_1_5km,
        turbulence_level_5_10km,
        solar_radiation_levels,
        geomagnetic_activity,
        solar_flare_activity,
        cosmic_ray_intensity,
        lightning_frequency,
        lightning_proximity,
        cloud_base_altitude,
        cloud_top_altitude,
        cloud_thickness,
        cloud_coverage_percentage,
        sea_surface_temperature,
        wave_height,
        wave_direction,
        current_speed,
        current_direction,
        pm2_5,
        pm10,
        co,
        o3,
        no2,
        so2,
        ground_temperature,
        air_temperature_0_1km,
        air_temperature_1_5km,
        air_temperature_5_10km,
        dew_point,
        ground_moisture_level,
        soil_temperature,
        fog_presence_and_density
    ) VALUES (
        %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s,%s,%s,%s,%s
    )
    """
    values = (
        launch_id,
        data["timestamp"],
        data["weather_conditions"]["surface_temperature"],
        data["weather_conditions"]["surface_wind_speed"],
        data["weather_conditions"]["surface_wind_direction"],
        data["weather_conditions"]["surface_humidity"],
        data["weather_conditions"]["precipitation_type"],
        data["weather_conditions"]["precipitation_intensity"],
        data["weather_conditions"]["visibility"],
        data["upper_atmosphere_conditions"]["upper_wind_speeds"]["0-1km"],
        data["upper_atmosphere_conditions"]["upper_wind_speeds"]["1-5km"],
        data["upper_atmosphere_conditions"]["upper_wind_speeds"]["5-10km"],
        data["upper_atmosphere_conditions"]["upper_wind_directions"]["0-1km"],
        data["upper_atmosphere_conditions"]["upper_wind_directions"]["1-5km"],
        data["upper_atmosphere_conditions"]["upper_wind_directions"]["5-10km"],
        data["upper_atmosphere_conditions"]["temperature_profiles"]["0-1km"],
        data["upper_atmosphere_conditions"]["temperature_profiles"]["1-5km"],
        data["upper_atmosphere_conditions"]["temperature_profiles"]["5-10km"],
        data["upper_atmosphere_conditions"]["humidity_profiles"]["0-1km"],
        data["upper_atmosphere_conditions"]["humidity_profiles"]["1-5km"],
        data["upper_atmosphere_conditions"]["humidity_profiles"]["5-10km"],
        data["upper_atmosphere_conditions"]["pressure_profiles"]["0-1km"],
        data["upper_atmosphere_conditions"]["pressure_profiles"]["1-5km"],
        data["upper_atmosphere_conditions"]["pressure_profiles"]["5-10km"],
        data["upper_atmosphere_conditions"]["turbulence_levels"]["0-1km"],
        data["upper_atmosphere_conditions"]["turbulence_levels"]["1-5km"],
        data["upper_atmosphere_conditions"]["turbulence_levels"]["5-10km"],
        data["solar_activity"]["solar_radiation_levels"],
        data["solar_activity"]["geomagnetic_activity"],
        data["solar_activity"]["solar_flare_activity"],
        data["solar_activity"]["cosmic_ray_intensity"],
        data["lightning_activity"]["lightning_frequency"],
        data["lightning_activity"]["lightning_proximity"],
        data["cloud_cover"]["cloud_base_altitude"],
        data["cloud_cover"]["cloud_top_altitude"],
        data["cloud_cover"]["cloud_thickness"],
        data["cloud_cover"]["cloud_coverage_percentage"],
        data["marine_conditions"]["sea_surface_temperature"],
        data["marine_conditions"]["wave_height"],
        data["marine_conditions"]["wave_direction"],
        data["marine_conditions"]["current_speed"],
        data["marine_conditions"]["current_direction"],
        data["air_quality"]["pm2_5"],
        data["air_quality"]["pm10"],
        data["air_quality"]["co"],
        data["air_quality"]["o3"],
        data["air_quality"]["no2"],
        data["air_quality"]["so2"],
        data["thermal_conditions"]["ground_temperature"],
        data["thermal_conditions"]["air_temperature_at_different_altitudes"]["0-1km"],
        data["thermal_conditions"]["air_temperature_at_different_altitudes"]["1-5km"],
        data["thermal_conditions"]["air_temperature_at_different_altitudes"]["5-10km"],
        data["thermal_conditions"]["dew_point"],
        data["launch_site_specific_conditions"]["ground_moisture_level"],
        data["launch_site_specific_conditions"]["soil_temperature"],
        data["launch_site_specific_conditions"]["fog_presence_and_density"]
    )
    cur.execute(query, values)
    conn.commit()

```

```python

# Retrieve all launch IDs and dates from the launch table
cur.execute("SELECT id, upcoming_launch_date FROM launch")
launch_data = cur.fetchall()


# Generate and insert data for each launch
for launch_id, date_of_launch in launch_data:
    base_date = datetime.datetime.combine(date_of_launch, datetime.time())  # Convert date to datetime
    weather_data = generate_random_data(base_date)
    insert_data(launch_id, weather_data)

# Commit the transaction and close the connection
conn.commit()
cur.close()
conn.close()
```
